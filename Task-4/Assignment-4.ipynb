### Question 1
Download a small dataset of IPL cricket player stats (e.g., runs, wickets, matches, strike rate) and apply StandardScaler from scikit-learn to scale all numeric features. Save the scaled data to a new CSV file named ipl_scaled.csv.

`StandardScaler` standardizes numeric features by removing the mean and scaling to unit variance ($z = \frac{x - \mu}{\sigma}$). The numeric columns (`matches`, `runs`, `wickets`, and `strike_rate`) are transformed so that each has a mean of approximately $0$ and a standard deviation of $1$, and the resulting DataFrame is exported to `ipl_scaled.csv`.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load IPL cricket player stats
df_ipl = pd.read_csv('ipl_cricket_player_stats_500.csv')
numeric_cols = ['matches', 'runs', 'wickets', 'strike_rate']

# Fit and transform numeric features using StandardScaler
scaler = StandardScaler()
df_ipl_scaled = df_ipl.copy()
df_ipl_scaled[numeric_cols] = scaler.fit_transform(df_ipl[numeric_cols])

# Save the scaled dataset to a new CSV file
df_ipl_scaled.to_csv('ipl_scaled.csv', index=False)

print("Scaled DataFrame preview (mean ~ 0, std ~ 1):")
display(df_ipl_scaled.head())
print("\nSummary statistics after scaling:")
print(df_ipl_scaled[numeric_cols].describe().round(3))
```

### Question 2
Using a dataset of Flipkart product ratings (features: price, rating, number of reviews, discount), apply MinMaxScaler so that all features are scaled between 0 and 1. Print the minimum and maximum value for each column before and after scaling.

*(Hint: Use pandas to load the data and scikit-learn's MinMaxScaler for scaling.)*

`MinMaxScaler` linearly transforms each numeric feature into the fixed range $[0, 1]$ using the formula $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$. This normalizes features having wide variations in scale (e.g., price in thousands vs. ratings out of 5) without distorting relative differences.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load Flipkart product ratings dataset
df_flipkart = pd.read_csv('flipkart_product_ratings_500.csv')
features = ['price', 'rating', 'number_of_reviews', 'discount']

# Inspect min and max before scaling
print("=== Min and Max BEFORE Scaling ===")
for col in features:
    print(f"{col:20s} -> Min: {df_flipkart[col].min():10.2f} | Max: {df_flipkart[col].max():10.2f}")

# Apply MinMaxScaler to scale features between 0 and 1
minmax = MinMaxScaler()
df_flipkart_scaled = df_flipkart.copy()
df_flipkart_scaled[features] = minmax.fit_transform(df_flipkart[features])

# Inspect min and max after scaling
print("\n=== Min and Max AFTER Scaling ===")
for col in features:
    print(f"{col:20s} -> Min: {df_flipkart_scaled[col].min():10.2f} | Max: {df_flipkart_scaled[col].max():10.2f}")
```

### Question 3
Given a Spotify songs dataset (features: danceability, energy, tempo, popularity), use correlation analysis to select the top 2 features most related to 'popularity'. List the selected features and explain your choice in one line each.

Correlation analysis via Pearson's correlation coefficient evaluates the linear relationship between acoustic attributes and `popularity` score. The top 2 selected features with the highest absolute correlation values are:

1. **`danceability`:** Songs with higher rhythmic regularity and strong beat patterns exhibit the strongest positive correlation with track streaming frequency and user popularity.
2. **`energy`:** Perceptual intensity and musical activity strongly drive listener engagement and playlist additions, making it the second most correlated predictor of popularity.

```python
import pandas as pd

# Load Spotify songs dataset
df_spotify = pd.read_csv('spotify_songs_correlation_500.csv')
cols = ['danceability', 'energy', 'tempo', 'popularity']

# Compute Pearson correlation matrix
corr_matrix = df_spotify[cols].corr()
popularity_corr = corr_matrix['popularity'].drop('popularity').abs().sort_values(ascending=False)

print("Full Correlation Matrix with Popularity:")
print(corr_matrix[['popularity']].round(4))

print("\nRanked by Absolute Correlation with Popularity:")
for feat, val in popularity_corr.items():
    print(f"{feat:15s}: {val:.4f}")

top_2 = popularity_corr.index[:2].tolist()
print(f"\nTop 2 Selected Features: {top_2}")
```

### Question 4
Take a Zomato restaurant dataset and use scikit-learn's RFE (Recursive Feature Elimination) with a DecisionTreeRegressor to select the 3 most important features for predicting 'average cost for two'. List the selected features and the RFE ranking.

Recursive Feature Elimination (RFE) fits a `DecisionTreeRegressor` and iteratively removes the least important features based on tree feature importance splits. The 3 selected features (ranked 1) are:

- **`has_table_booking` (Rank 1):** Fine-dining restaurants providing table reservations have substantially higher average price tiers.
- **`votes` (Rank 1):** Reviewer volume strongly separates high-end, frequented culinary establishments from small local eateries.
- **`cuisines_count` (Rank 1):** Multi-cuisine menus require extensive kitchen operations, correlating strongly with increased dining costs.

```python
import pandas as pd
from sklearn.feature_selection import RFE
from sklearn.tree import DecisionTreeRegressor

# Load Zomato restaurant dataset
df_zomato = pd.read_csv('zomato_restaurants_500.csv')
feature_cols = ['has_table_booking', 'has_online_delivery', 'votes', 'aggregate_rating', 'cuisines_count']
target_col = 'average_cost_for_two'

X = df_zomato[feature_cols]
y = df_zomato[target_col]

# Apply RFE with DecisionTreeRegressor to select 3 features
estimator = DecisionTreeRegressor(random_state=42)
rfe = RFE(estimator=estimator, n_features_to_select=3)
rfe.fit(X, y)

# Compile RFE ranking table
rfe_results = pd.DataFrame({
    'Feature': feature_cols,
    'Selected': rfe.support_,
    'Ranking': rfe.ranking_
}).sort_values('Ranking')

print("RFE Feature Selection Ranking:")
display(rfe_results)

selected_features = rfe_results[rfe_results['Selected']]['Feature'].tolist()
print(f"Selected Top 3 Features: {selected_features}")
```

### Question 5
Given a BookMyShow movie ratings dataset, detect outliers in the 'user ratings' column using the IQR method. Remove the outliers and display how many rows were dropped.

*(Hint: Calculate Q1, Q3, and IQR; filter out rows outside [Q1 - 1.5*IQR, Q3 + 1.5*IQR].)*

The Interquartile Range (IQR) method defines the middle 50% spread ($IQR = Q_3 - Q_1$). Ratings falling below $Q_1 - 1.5 \times IQR$ or above $Q_3 + 1.5 \times IQR$ represent abnormal scores (such as data scraping anomalies or extreme rating sabotage) and are pruned to ensure a clean distribution.

```python
import pandas as pd

# Load BookMyShow movie ratings dataset
df_bms = pd.read_csv('bookmyshow_ratings_500.csv')
initial_rows = len(df_bms)

# Calculate Q1 (25th percentile), Q3 (75th percentile), and IQR
q1 = df_bms['user_ratings'].quantile(0.25)
q3 = df_bms['user_ratings'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Q1 (25th percentile): {q1:.2f}")
print(f"Q3 (75th percentile): {q3:.2f}")
print(f"IQR:                 {iqr:.2f}")
print(f"Valid Range:         [{lower_bound:.2f}, {upper_bound:.2f}]")

# Filter out outliers outside the [lower_bound, upper_bound] interval
df_cleaned = df_bms[(df_bms['user_ratings'] >= lower_bound) & (df_bms['user_ratings'] <= upper_bound)]
rows_dropped = initial_rows - len(df_cleaned)

print(f"\nInitial rows:        {initial_rows}")
print(f"Rows after cleaning: {len(df_cleaned)}")
print(f"Total rows dropped:  {rows_dropped}")
```
