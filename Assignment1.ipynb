### Question 3
Draw a simple diagram (hand-drawn or digital) showing the basic ML workflow: Data → Preprocessing → Modeling → Evaluation → Deployment. Under each step, write a one-line example of what happens at that stage for a food delivery app like Swiggy.

```
+----------------+      +-----------------+      +---------------+      +----------------+      +----------------+
|      Data      | ---> |  Preprocessing  | ---> |   Modeling    | ---> |   Evaluation   | ---> |   Deployment   |
+----------------+      +-----------------+      +---------------+      +----------------+      +----------------+
```

- **Data:** Collecting raw order logs containing delivery partner GPS tracks, restaurant food prep durations, weather data, and real-time traffic conditions.
- **Preprocessing:** Handling missing sensor readings, standardizing distance units, encoding categorical weather types, and normalizing outlier prep times.
- **Modeling:** Training a Gradient Boosted Decision Tree (or Deep Neural Network) to regress estimated time of delivery (ETA) in minutes.
- **Evaluation:** Calculating Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE) against actual historical delivery completion timestamps on held-out test data.
- **Deployment:** Serving the trained ETA model via low-latency REST APIs to return live arrival estimates on customer checkout screens.

````python
# Code representation: End-to-end ML Workflow demonstration for Swiggy ETA prediction
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Data Collection (Simulated: Distance in km, Kitchen prep in mins, Traffic factor 1-3)
np.random.seed(42)
X_raw = np.column_stack([
    np.random.uniform(1.0, 15.0, 100),  # Distance (km)
    np.random.uniform(10.0, 30.0, 100), # Prep time (mins)
    np.random.randint(1, 4, 100)        # Traffic index (1: Low, 2: Med, 3: High)
])
# True ETA target = prep_time + (distance * 3 * traffic_index) + noise
y = X_raw[:, 1] + (X_raw[:, 0] * 2.5 * X_raw[:, 2]) + np.random.normal(0, 2, 100)

# 2. Preprocessing
X_train, X_test, y_train, y_test = train_test_split(X_raw, y, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Modeling
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# 4. Evaluation
y_pred = model.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
print(f"Evaluation - Mean Absolute Error (MAE): {mae:.2f} minutes")

# 5. Deployment (Live prediction inference function)
def live_predict_eta(distance_km, prep_time_min, traffic_index):
    sample = np.array([[distance_km, prep_time_min, traffic_index]])
    scaled_sample = scaler.transform(sample)
    eta_estimate = model.predict(scaled_sample)[0]
    return round(float(eta_estimate), 1)

sample_eta = live_predict_eta(distance_km=5.2, prep_time_min=15.0, traffic_index=2)
print(f"Deployment API Output: Estimated Delivery Time = {sample_eta} minutes")
````

### Question 4
Imagine you are building a spam filter for WhatsApp messages. Write a short paragraph explaining how traditional programming would approach this problem versus how machine learning would approach it.

*(Hint: Focus on rules vs learning from data.)*

In **traditional programming**, a human developer manually writes explicit, deterministic `if-else` rules and regular expressions to block messages containing predefined keywords like `"congratulations"`, `"lottery"`, `"claim prize"`, or suspicious phone number patterns. While simple to implement, this approach is rigid and rapidly fails whenever spammers make minor spelling alterations (e.g., `"w!n cash"` or `"l0ttery"`), requiring ongoing manual maintenance of an exhaustive rules list.

In contrast, a **machine learning approach** does not rely on hardcoded conditional rules; instead, an algorithm (such as Naive Bayes, Logistic Regression, or a Transformer) is trained on thousands of labeled historical messages marked as *Spam* or *Ham (Legitimate)*. The ML model automatically learns probabilistic token associations, contextual semantic patterns, and sender metadata, generalizing to novel variations and evolving spamming tactics autonomously without manual rule updates.

````python
# Comparison demonstration: Rule-based Filter vs. ML Classifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Rule-Based Filter Implementation
def rule_based_spam_filter(text):
    banned_words = ["lottery", "prize", "free cash", "won", "claim"]
    return "Spam" if any(word in text.lower() for word in banned_words) else "Ham"

# 2. Machine Learning Classifier Implementation
train_messages = [
    "Hey, are we still meeting for lunch?",
    "Claim your free lottery prize now!",
    "Can you review the attached document?",
    "Congratulations you won cash rewards immediately",
    "Let me know when you arrive home.",
    "Exclusive offer: win big cash payout today"
]
labels = ["Ham", "Spam", "Ham", "Spam", "Ham", "Spam"]

vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(train_messages)
ml_classifier = MultinomialNB()
ml_classifier.fit(X_train_vec, labels)

# Test on a variation that bypasses simple exact keyword matching
incoming_msg = "Congratulations! Click to claim rewards today"

print(f"Message: '{incoming_msg}'")
print("Rule-based detection:", rule_based_spam_filter(incoming_msg))
sample_vec = vectorizer.transform([incoming_msg])
print("ML Classifier detection:", ml_classifier.predict(sample_vec)[0])
````
