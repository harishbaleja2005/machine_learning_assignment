### Question 1
Download a small sample of IPL match data (CSV) with some missing values in player stats. Load it using pandas and print the number of missing values in each column.

The dataset contains 500 rows of IPL player match statistics. Using pandas' `isnull().sum()` method identifies missing values across all numerical and categorical columns.

```python
import pandas as pd

# Load the IPL player match data
df = pd.read_csv('ipl_player_match_data_500.csv')

# Inspect dataset dimensions and count missing values in each column
print(f"Total rows and columns: {df.shape}")
print("\nMissing values count per column:")
print(df.isnull().sum())
```

### Question 2
For the 'player_age' column in your IPL dataset, fill all missing values with the median age using pandas' fillna() method and display the updated column.

The median age is robust against potential skewness and outliers. Imputing missing ages using `fillna(median_age)` ensures complete numerical data while retaining the column's central tendency.

```python
# Calculate the median age excluding NaN values
median_age = df['player_age'].median()
print(f"Calculated Median Age: {median_age}")

# Impute missing values with the median age
df['player_age'] = df['player_age'].fillna(median_age)

print(f"Missing values in 'player_age' after imputation: {df['player_age'].isnull().sum()}")
print("\nUpdated 'player_age' column (first 10 entries):")
print(df['player_age'].head(10))
```

### Question 3
Suppose the 'team' column in your IPL dataset contains missing values. Replace all missing team names with the constant value 'Unknown' and print the first 10 rows.

Missing categorical entries in the `team` column are imputed with the string constant `'Unknown'` using `fillna('Unknown')`, preserving row availability without biasing specific team tallies.

```python
# Impute missing values in 'team' column with 'Unknown'
df['team'] = df['team'].fillna('Unknown')

print(f"Missing values in 'team' after replacement: {df['team'].isnull().sum()}")
print("\nFirst 10 rows of the updated DataFrame:")
display(df.head(10))
```

### Question 4
Take the 'venue' column (categorical) from your IPL dataset and apply one-hot encoding using pandas' get_dummies(). Show the resulting DataFrame with the new columns.

One-hot encoding converts the nominal categorical `venue` column into distinct binary (0 or 1) columns for each unique stadium, making it suitable for mathematical models without imposing an artificial ordinal ranking.

```python
# One-hot encode the 'venue' column
venue_dummies = pd.get_dummies(df['venue'], prefix='venue', dtype=int)

# Concatenate dummy columns with the main DataFrame
df_encoded = pd.concat([df.drop(columns=['venue']), venue_dummies], axis=1)

print(f"Generated one-hot columns: {list(venue_dummies.columns)}")
print(f"New DataFrame shape: {df_encoded.shape}")
print("\nPreview of DataFrame with one-hot encoded venue columns:")
display(df_encoded[['player_name', 'team'] + list(venue_dummies.columns)].head(5))
```

### Question 5
Use LabelEncoder from sklearn to encode the 'player_role' column (e.g., Batsman, Bowler, Allrounder) in your IPL dataset, and display the mapping from original roles to encoded values.

*(Hint: Use LabelEncoder's classes_ attribute to see the mapping.)*

`LabelEncoder` assigns each unique category string an integer from $0$ to $k-1$ in alphabetical order. The `classes_` attribute stores the mapped labels in index order.

```python
from sklearn.preprocessing import LabelEncoder

# Instantiate and fit LabelEncoder on 'player_role'
le = LabelEncoder()
df['player_role_encoded'] = le.fit_transform(df['player_role'])

# Display the mapping from original class strings to encoded integer values
role_mapping = {label: index for index, label in enumerate(le.classes_)}

print("LabelEncoder Mapping (Role -> Integer):")
for role, code in role_mapping.items():
    print(f"  {role:15s} -> {code}")

print("\nComparison of original role and encoded values:")
display(df[['player_name', 'player_role', 'player_role_encoded']].head(8))
```
